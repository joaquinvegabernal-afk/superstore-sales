Limpieza de datos

Las columnas Order Date y Ship Date vienen como texto. Se deben  convertir  a objetos de fecha (datetime).
Limpieza de Strings: Se revisara si hay espacios en blanco al principio o al final de los nombres de ciudades o categorías y quítalos.

```python
#Convertimos nuestras columnas de texto a fechas
data['Order Date'] = pd.to_datetime( data["Order Date"], format= "%d-%m-%Y" ,errors= 'coerce')
data['Ship Date'] = pd.to_datetime( data["Ship Date"], format= "%d-%m-%Y",errors= 'coerce')
```

```python
data.head()
```

```python
#Verificamos que se cambio el tipo de dato de Order date y Ship date
data.info()
```

```python
#Verificando si tenemos datos vacios o nulos
data.isnull().sum()
```

```python
#Eliminando filas repetidas 
print(f"Tamaño del set antes de eliminar filas duplicadas {data.shape}")
data.drop_duplicates(inplace= True)
print(f"Tamaño del set despues de eliminar filas duplicadas {data.shape}")
```

```python
#Quitando espacios en blanco en las columans string
columnas_a_limpiar = ["Order ID","Ship Mode","Customer ID","Customer Name","Segment","Country","City","State","Region","Product ID","Category","Sub-Category","Product Name"]
data[columnas_a_limpiar] = data[columnas_a_limpiar].apply(lambda x : x.str.strip())
```

Nuevas Métricas

Tiempo de Entrega: Crearemos una columna que calcule la diferencia en días entre la fecha de envío (ship date) y la fecha de la orden (Order Date).

Costo Total : Costo total de nuestrto dataframe

Porcentaje de Margen: Se calculara qué porcentaje representa la utilidad respecto a la venta para cada fila.


```python
data["Porcentaje de margen"] = (data["Profit"] / data["Sales"]) * 100
```

```python
data["Tiempo de entrega"] = data["Ship Date"] - data["Order Date"]
```

```python
data["Costo total"] = (data["Sales"] - data["Profit"]).round(2)
```
